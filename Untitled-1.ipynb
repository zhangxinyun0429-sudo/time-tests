```python
import numpy as np
import matplotlib.pyplot as plt
import numba
from numba import cuda
import math
import time

# ============================================================================
# Part 1: CPU Implementation
# ============================================================================

def solve_heat_cpu(N, t_max):
    """
    Solve the heat equation on CPU
    
    Parameters:
    -----------
    N : int
        Number of spatial grid points (including boundaries)
    t_max : float
        Maximum time to simulate
    
    Returns:
    --------
    u : ndarray
        Solution at time t_max
    x : ndarray
        Spatial grid points
    """
    h = 1.0 / N  # spatial and temporal step size
    n_steps = int(t_max / h)
    
    # Initialize spatial grid
    x = np.linspace(0, 1, N + 1)
    
    # Initialize solution: u = 0 everywhere except boundaries
    u_current = np.zeros(N + 1)
    u_current[0] = 10.0
    u_current[N] = 10.0
    
    u_next = np.zeros(N + 1)
    
    # Time stepping
    for step in range(n_steps):
        # Boundary conditions
        u_next[0] = 10.0
        u_next[N] = 10.0
        
        # Interior points
        for i in range(1, N):
            u_next[i] = u_current[i] + (u_current[i-1] - 2*u_current[i] + u_current[i+1]) / (1000 * h)
        
        # Swap arrays
        u_current, u_next = u_next, u_current
    
    return u_current, x

# ============================================================================
# Part 2: GPU Implementation with CUDA
# ============================================================================

@cuda.jit
def heat_step_kernel(u_current, u_next, N, h):
    """
    CUDA kernel for one time step of heat equation
    Each thread handles one interior point
    """
    i = cuda.grid(1)
    
    # Check bounds (only process interior points)
    if 1 <= i < N:
        u_next[i] = u_current[i] + (u_current[i-1] - 2*u_current[i] + u_current[i+1]) / (1000 * h)

def solve_heat_gpu(N, t_max, return_all_steps=False):
    """
    Solve the heat equation on GPU
    
    Parameters:
    -----------
    N : int
        Number of spatial grid points (including boundaries)
    t_max : float
        Maximum time to simulate
    return_all_steps : bool
        If True, return solution at all time steps
    
    Returns:
    --------
    u : ndarray
        Solution at time t_max (or all time steps if return_all_steps=True)
    x : ndarray
        Spatial grid points
    times : ndarray (optional)
        Time points if return_all_steps=True
    """
    h = 1.0 / N
    n_steps = int(t_max / h)
    
    # Initialize on CPU
    x = np.linspace(0, 1, N + 1)
    u_current = np.zeros(N + 1, dtype=np.float64)
    u_current[0] = 10.0
    u_current[N] = 10.0
    u_next = np.zeros(N + 1, dtype=np.float64)
    u_next[0] = 10.0
    u_next[N] = 10.0
    
    # Copy to GPU (only copy once!)
    d_u_current = cuda.to_device(u_current)
    d_u_next = cuda.to_device(u_next)
    
    # Configure kernel
    threads_per_block = 256
    blocks_per_grid = (N + threads_per_block - 1) // threads_per_block
    
    # Store history if needed
    if return_all_steps:
        u_history = []
        time_points = []
    
    # Time stepping on GPU
    for step in range(n_steps):
        # Launch kernel
        heat_step_kernel[blocks_per_grid, threads_per_block](d_u_current, d_u_next, N, h)
        
        # Swap device arrays (just swap references, no copy!)
        d_u_current, d_u_next = d_u_next, d_u_current
        
        # Store history if needed
        if return_all_steps:
            if step % max(1, n_steps // 1000) == 0:  # Store every ~1000th step
                u_history.append(d_u_current.copy_to_host())
                time_points.append((step + 1) * h)
    
    # Copy final result back to CPU (only copy once at the end!)
    u_final = d_u_current.copy_to_host()
    
    if return_all_steps:
        return u_final, x, np.array(time_points), u_history
    else:
        return u_final, x

# ============================================================================
# Part 3: Choose sensible N and plot solutions
# ============================================================================

def choose_N_and_plot():
    """
    Choose a sensible value of N and plot solutions at t=1, 2, 10
    
    Reasoning for N choice:
    - We need N large enough for accuracy but not too large for computational cost
    - The stability condition for explicit schemes is typically h < 1/2 for stability
    - Since h = 1/N, we need N to be reasonably large
    - N = 1000 gives h = 0.001, which should be stable and accurate
    """
    N = 1000
    print(f"Chosen N = {N:,}")
    print(f"Spatial step size h = {1/N:.6f}")
    print(f"Time step size = h = {1/N:.6f}")
    print(f"\nReasoning: N=1000 provides good balance between accuracy and")
    print(f"computational cost. The step size h=0.001 is small enough for")
    print(f"stability and accuracy of the explicit time-stepping scheme.")
    
    # Solve at different times
    times_to_plot = [1.0, 2.0, 10.0]
    
    plt.figure(figsize=(14, 5))
    
    for idx, t in enumerate(times_to_plot):
        print(f"\nSolving for t = {t}...")
        start = time.time()
        u, x = solve_heat_gpu(N, t)
        elapsed = time.time() - start
        print(f"  Time taken: {elapsed:.4f} seconds")
        print(f"  Temperature at midpoint: {u[N//2]:.6f}")
        
        plt.subplot(1, 3, idx + 1)
        plt.plot(x, u, 'b-', linewidth=2)
        plt.xlabel('Position x', fontsize=12)
        plt.ylabel('Temperature u', fontsize=12)
        plt.title(f'Temperature at t = {t}', fontsize=14, fontweight='bold')
        plt.grid(True, alpha=0.3)
        plt.ylim([0, 11])
        plt.axhline(y=9.8, color='r', linestyle='--', alpha=0.5, label='Target temp = 9.8')
        if t == times_to_plot[-1]:
            plt.legend()
    
    plt.tight_layout()
    plt.savefig('heat_equation_solution.png', dpi=300, bbox_inches='tight')
    plt.show()
    print("\nPlot saved as 'heat_equation_solution.png'")

# ============================================================================
# Part 4: Find time when midpoint exceeds 9.8
# ============================================================================

def find_time_exceeds_threshold(N=1000, threshold=9.8):
    """
    Find the time when temperature at midpoint first exceeds threshold
    
    Strategy:
    1. First, do a coarse search with large time steps to bracket the answer
    2. Then, do a fine search in the bracketed interval
    3. Use binary search for efficiency
    """
    print(f"\n{'='*70}")
    print(f"Finding time when midpoint temperature exceeds {threshold}")
    print(f"{'='*70}")
    
    h = 1.0 / N
    midpoint_idx = N // 2
    
    # Phase 1: Coarse search to bracket the time
    print(f"\nPhase 1: Coarse search...")
    t_coarse = 0.0
    dt_coarse = 0.5  # Check every 0.5 time units
    
    while True:
        u, x = solve_heat_gpu(N, t_coarse)
        temp_mid = u[midpoint_idx]
        print(f"  t = {t_coarse:.2f}, temp at midpoint = {temp_mid:.6f}")
        
        if temp_mid >= threshold:
            t_upper = t_coarse
            t_lower = t_coarse - dt_coarse
            print(f"\n  Found bracket: [{t_lower:.2f}, {t_upper:.2f}]")
            break
        
        t_coarse += dt_coarse
        
        if t_coarse > 20:  # Safety check
            print("  Warning: exceeded t=20, stopping search")
            return None
    
    # Phase 2: Binary search for precise time
    print(f"\nPhase 2: Binary search for precise time...")
    tolerance = 0.01  # Find time within 0.01 units
    
    while (t_upper - t_lower) > tolerance:
        t_mid = (t_lower + t_upper) / 2
        u, x = solve_heat_gpu(N, t_mid)
        temp_mid = u[midpoint_idx]
        
        if temp_mid < threshold:
            t_lower = t_mid
        else:
            t_upper = t_mid
        
        print(f"  t ∈ [{t_lower:.4f}, {t_upper:.4f}], temp at t_mid = {temp_mid:.6f}")
    
    t_result = (t_lower + t_upper) / 2
    
    # Verify result
    u_final, x = solve_heat_gpu(N, t_result)
    temp_final = u_final[midpoint_idx]
    
    print(f"\n{'='*70}")
    print(f"RESULT")
    print(f"{'='*70}")
    print(f"Time when midpoint exceeds {threshold}: t ≈ {t_result:.4f}")
    print(f"Temperature at midpoint at this time: {temp_final:.6f}")
    print(f"\nMethod: Used binary search to find the precise time.")
    print(f"First did coarse search with Δt=0.5 to bracket the answer,")
    print(f"then used binary search to refine to within tolerance of 0.01.")
    
    # Create visualization
    create_threshold_plot(N, t_result, threshold)
    
    return t_result

def create_threshold_plot(N, t_critical, threshold=9.8):
    """
    Create a plot showing temperature evolution at midpoint
    """
    print(f"\nCreating visualization...")
    
    # Sample times around the critical time
    t_samples = np.linspace(0, t_critical + 1, 50)
    temps_midpoint = []
    
    midpoint_idx = N // 2
    
    for t in t_samples:
        u, x = solve_heat_gpu(N, t)
        temps_midpoint.append(u[midpoint_idx])
    
    # Also get spatial profile at critical time
    u_critical, x = solve_heat_gpu(N, t_critical)
    
    # Create figure with two subplots
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
    
    # Left plot: Temperature at midpoint vs time
    ax1.plot(t_samples, temps_midpoint, 'b-', linewidth=2, label='Midpoint temperature')
    ax1.axhline(y=threshold, color='r', linestyle='--', linewidth=2, label=f'Threshold = {threshold}')
    ax1.axvline(x=t_critical, color='g', linestyle=':', linewidth=2, alpha=0.7, 
                label=f't = {t_critical:.4f}')
    ax1.plot([t_critical], [threshold], 'ro', markersize=12, label='Critical point')
    ax1.set_xlabel('Time t', fontsize=12, fontweight='bold')
    ax1.set_ylabel('Temperature at midpoint', fontsize=12, fontweight='bold')
    ax1.set_title('Temperature Evolution at Midpoint', fontsize=14, fontweight='bold')
    ax1.legend(fontsize=10)
    ax1.grid(True, alpha=0.3)
    
    # Right plot: Spatial profile at critical time
    ax2.plot(x, u_critical, 'b-', linewidth=2, label=f'Temperature at t={t_critical:.4f}')
    ax2.axhline(y=threshold, color='r', linestyle='--', linewidth=2, alpha=0.5, 
                label=f'Threshold = {threshold}')
    ax2.plot([0.5], [u_critical[N//2]], 'ro', markersize=12, label='Midpoint')
    ax2.set_xlabel('Position x', fontsize=12, fontweight='bold')
    ax2.set_ylabel('Temperature', fontsize=12, fontweight='bold')
    ax2.set_title(f'Temperature Profile at t = {t_critical:.4f}', fontsize=14, fontweight='bold')
    ax2.legend(fontsize=10)
    ax2.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.savefig('threshold_analysis.png', dpi=300, bbox_inches='tight')
    plt.show()
    print("Plot saved as 'threshold_analysis.png'")

# ============================================================================
# Part 5: Performance comparison and timing
# ============================================================================

def compare_cpu_gpu_performance():
    """
    Compare CPU and GPU performance
    """
    print(f"\n{'='*70}")
    print("PERFORMANCE COMPARISON: CPU vs GPU")
    print(f"{'='*70}")
    
    N_values = [100, 500, 1000, 2000]
    t_test = 2.0
    
    cpu_times = []
    gpu_times = []
    
    print(f"\nTesting with t_max = {t_test}")
    print(f"\n{'N':<10} {'CPU Time (s)':<15} {'GPU Time (s)':<15} {'Speedup':<10}")
    print("-" * 55)
    
    for N in N_values:
        # CPU timing
        start = time.time()
        u_cpu, x = solve_heat_cpu(N, t_test)
        cpu_time = time.time() - start
        cpu_times.append(cpu_time)
        
        # GPU timing
        start = time.time()
        u_gpu, x = solve_heat_gpu(N, t_test)
        gpu_time = time.time() - start
        gpu_times.append(gpu_time)
        
        speedup = cpu_time / gpu_time
        
        print(f"{N:<10} {cpu_time:<15.6f} {gpu_time:<15.6f} {speedup:<10.2f}x")
        
        # Verify results match
        error = np.max(np.abs(u_cpu - u_gpu))
        if error > 1e-10:
            print(f"  Warning: CPU and GPU results differ by {error:.2e}")
    
    # Plot performance
    plt.figure(figsize=(10, 6))
    plt.plot(N_values, cpu_times, 'bo-', linewidth=2, markersize=8, label='CPU')
    plt.plot(N_values, gpu_times, 'go-', linewidth=2, markersize=8, label='GPU')
    plt.xlabel('N', fontsize=12, fontweight='bold')
    plt.ylabel('Time (seconds)', fontsize=12, fontweight='bold')
    plt.title('CPU vs GPU Performance', fontsize=14, fontweight='bold')
    plt.legend(fontsize=11)
    plt.grid(True, alpha=0.3)
    plt.yscale('log')
    plt.xscale('log')
    plt.savefig('performance_comparison.png', dpi=300, bbox_inches='tight')
    plt.show()
    print("\nPerformance comparison plot saved as 'performance_comparison.png'")

# ============================================================================
# Main execution
# ============================================================================

if __name__ == "__main__":
    print("="*70)
    print("HEAT EQUATION SOLVER WITH GPU ACCELERATION")
    print("="*70)
    
    # Check if CUDA is available
    try:
        print(f"\nCUDA available: {cuda.is_available()}")
        if cuda.is_available():
            print(f"CUDA devices: {cuda.gpus}")
    except:
        print("\nWarning: Could not check CUDA availability")
    
    # Part 1: Choose N and plot solutions
    print("\n" + "="*70)
    print("PART 1: Choosing N and plotting solutions")
    print("="*70)
    choose_N_and_plot()
    
    # Part 2: Find time when midpoint exceeds 9.8
    print("\n" + "="*70)
    print("PART 2: Finding critical time")
    print("="*70)
    t_critical = find_time_exceeds_threshold(N=1000, threshold=9.8)
    
    # Part 3: Performance comparison (optional)
    print("\n" + "="*70)
    print("PART 3: Performance comparison (optional)")
    print("="*70)
    response = input("\nDo you want to run CPU vs GPU performance comparison? (y/n): ")
    if response.lower() == 'y':
        compare_cpu_gpu_performance()
    
    print("\n" + "="*70)
    print("ALL TASKS COMPLETED")
    print("="*70)
```
